# Aula - Manipulação Estrutural com Pandas

Manipulação estrutural diz respeito à modificação da estrutura de uma dataframe. Isso é realizado na etapa de ETL.

**Um dataframe é composto de:** <span style="color: red">Colunas</span> e linhas.

**A estrutura de uma dataframe é determinada pela configuração das suas <span style="color: red">colunas</span>.**
- Quais colunas fazem parte do dataframe?
- Quais os tipos dessas colunas?
- Quais os nomes dessas colunas?

**A manipulação de estrutura é:**
- Definir quais colunas farão parte do df
- Remover colunas de um df
- Criar novas colunas
- Renomear uma coluna existente
- Alterar o tipo de uma coluna existente
- Modificar o index de um dataframe

```python
import pandas as pd # importando biblioteca Pandas
```

### Carregando o Dataframe

```python
df = pd.read_excel("./resultados.xlsx") # carregando base de dados
df[[
    "Edition", "Round", "Date", "Home Club", "Away Club", "Home Score", "AwayScore" # criando Novas colunas
]] = df["Edition,Round,Date,Home Club,Away Club,Home Score,AwayScore"].str.split(",", expand=True)

df
```

### Definir quais colunas farão parte do df

```python
df_select_columns = df
df_select_columns = df[["Edition", "Date", "Home Club", "Away Club", "Home Score", "AwayScore"]] # definindo as colunas que fara parte da base de dados
df_select_columns
```

### Removendo colunas de um dataframe

```python
df_remove_columns = df_select_columns
df_remove_columns = df_remove_columns.drop(columns=["Edition"], axis=1) # removendo coluna da base de dados
df_remove_columns
```

### Adicionando novas colunas ao df

```python
df_add_columns = df_remove_columns
df_add_columns["Total Gols"] = df_add_columns["Home Score"] + df_add_columns["AwayScore"] # adicionando nova coluna na base de dados

df_add_columns
```

### Alterando o tipo de uma coluna existente

```python
df_types = df_add_columns

df_types["Home Score"] = df_types["Home Score"].astype(float) # alterando a tipagem de uma coluna
df_types["AwayScore"] = df_types["AwayScore"].astype(float)

df_types["Total Gols"] = df_types["Home Score"] + df_types["AwayScore"]
df_types
```

### Renomeando colunas de um dataframe

```python
df_rename = df_types
df_rename = df_rename.rename( # Renomeando Colunas da base de dados
    columns={
        "Date": "partida_data",
        "Home Club":  "time_mandante_nome",
        "Away Club":  "time_visitante_nome",
        "Home Score":  "time_mandante_gols",
        "AwayScore":  "time_visitante_gols",
        "Total Gols":  "partida_gols",
    }
)

df_rename
```

### Modificando o index de um df

```python
df_index = df_rename
df_index = df_index["partida_gols"] # Ele funciona exatamente como as chaves de um dicionário do Python , permitindo que você encontre, alinhe e manipule dados de forma rápida e eficiente.
df_index = df_index.reset_index()
df_index = df_index.set_index("index")
df_index
```
